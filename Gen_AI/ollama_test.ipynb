```python
from langchain_ollama import ChatOllama


llm = ChatOllama(
    model="qwen2.5:3b",
    temperature=1,
    top_k=10,
    top_p=0.9,
    num_predict=50,
    repeat_penalty=1.1,
    num_ctx=100,
    #seed=4
)

response = llm.invoke("Answer in English, Describe about Tajmahal?")

print(response.content)
```
