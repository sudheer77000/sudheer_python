```python
import os
from dotenv import load_dotenv
from groq import Groq
load_dotenv()
```

```python
client = Groq(api_key=os.getenv("GROQ_API_KEY"))
def ask_llm(prompt):
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role":"system","content":"You are a helpful AI teacher, explain in 2 sentences"},
            {"role":"user","content":prompt}
        ],
        temperature= 0.3
    )
    return response.choices[0].message.content

print(ask_llm("Explain About RKT Airport,UAE"))
```

```python
def calculator(expression):
    try:
        return eval(expression)
    except Exception as e:
        return f"Error: {e}"
def simple_agent(user_input):
    if "calculate" in user_input.lower():
        expression = user_input.lower().replace("calculate", "").strip()
        return calculator(expression)
    else:
        return ask_llm(user_input)

#print(simple_agent("calculate 23 * 9"))
print(simple_agent("About AUH Airport in 10 words"))
print(calculator("23 + 23"))
```

```python
from ddgs import DDGS

def web_search(query,max_results = 3):
    results = []
    with DDGS() as ddgs:
        for r in ddgs.text(query,max_results=max_results):
            results.append({
                "title":r.get("title"),
                "href":r.get("href"),
                "body":r.get("body"),
            })
    return results

query = "latest cricket news"
search_results = web_search(query)

for item in search_results:
    print(item["title"])
    print(item["href"])
    print(item["body"])
    print("-" * 50)
```

```python
def score_check_agent(user_query):
    search_data = web_search(f"{user_query}",max_results=3)

    context = "\n\n".join([
        f"Titla: {item['title']}\nSummery: {item['body']}\nURL : {item['href']}"
        for item in search_data
    ])
    print(context)

    prompt = f"""
    You are an Online Price comparision analyst

    Qyery Data :
    {user_query}

    web research data:
    {context}

    Give:
    1. Average Price Of the Product
    2. Available Online stores
    3. Rating
    
    """

    return ask_llm(prompt)

query = "Live cricket Score"
print(score_check_agent(query))
```
