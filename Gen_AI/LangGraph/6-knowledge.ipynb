```python
def add(a: int, b: int):
    return a + b

result = add(10, 20)
print("Addition Of Two Numbers : ",result)
```

```python
from langchain_core.tools import tool

@tool
def add_tool(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b
```

```python
result = add_tool.invoke({"a":10,"b":30})
print("Addition Of Two Numbers : ",result)
```

```python
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
```

```python
# Tool With Custom Name
@tool(
    "Addition_Tool",
    description="Adding two integers and return the result.",
    return_direct=False,
)
def add_tool(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b
```

```python
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
print(add_tool.args_schema.model_json_schema())
```

```python
import asyncio

@tool
async def get_customer(customer_id: int) -> str:
    """Get customer information."""

    # Imagine this takes 3 seconds
    print("**" * 32)
    await asyncio.sleep(3)

    return f"Customer {customer_id} details"
```

```python
results = await asyncio.gather(
    get_customer.ainvoke({"customer_id": 101}),
    get_customer.ainvoke({"customer_id": 102}),
    get_customer.ainvoke({"customer_id": 103})
)

print(results)
```
