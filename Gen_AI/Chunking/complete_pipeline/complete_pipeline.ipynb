```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text")
vector = embeddings.embed_query("How many years Ishitha Gundra was?")

print(type(vector))
print(len(vector))
print(vector[:10])
```

```python
chunks = [
    "Ishitha Gundra is a bright 9-year-old currently exploring foundational Grade 3 and Grade 4 subjects, including mathematics, English, and data handling.",
    "Ishitha Gundra enjoys creative and playful activities, with a special appreciation for fun, custom animal songs.",
    "As a cherished daughter, Ishitha Gundra benefits from a strongly supportive environment that actively encourages her daily learning and curiosity."
]

vectors = embeddings.embed_documents(chunks)

print(len(vectors))
print(len(vectors[0]))
#print(vectors)
```

```python
from langchain_chroma import Chroma
from langchain_core.documents import Document

docs =[
    Document(page_content="Ishitha Gundra is a bright 9-year-old currently exploring foundational Grade 3 and Grade 4 subjects, including mathematics, English, and data handling."),
    Document(page_content="Ishitha Gundra enjoys creative and playful activities, with a special appreciation for fun, custom animal songs."),
    Document(page_content="As a cherished daughter, Ishitha Gundra benefits from a strongly supportive environment that actively encourages her daily learning and curiosity.")
]

vectorStore = Chroma.from_documents(
    documents = docs,
    embedding=embeddings
)
print(vectorStore)
```

```python
results = vectorStore.similarity_search("How many years Ishitha Gundra was?",k=2)

for doc in results:
    print(doc.page_content)
```
