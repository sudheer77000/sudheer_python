```python
# pi chart using Matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
iplTeams = ['CSK', 'MI', 'RCB', 'KKR', 'SRH', 'DC', 'RR', 'PBKS']
trophies = [4, 5, 1, 2, 1, 3, 2, 4]
colors = ['yellow', 'blue', 'red', 'purple', 'orange', 'green', 'cyan', 'magenta']
plt.pie(trophies, labels=iplTeams, autopct='%1.1f%%', colors=colors, wedgeprops={'edgecolor': 'black'},shadow=True, explode=[0, 0, 0.3, 0, 0, 0, 0, 0])
plt.show()
```
