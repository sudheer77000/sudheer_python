```python
class BankAccount:
    def __init__(this, account_number, balance=0):
        this.account_number = account_number
        this.balance = balance
    def print_account_info(this):
        print(f"Account Number: {this.account_number}, Balance: {this.balance}")
    def deposit(this, amount):
        this.balance += amount
    def withdraw(this, amount):
        if amount > this.balance:
            print("Insufficient funds")
        else:
            this.balance -= amount
    def check_balance(this):
        return this.balance 
BAObject1 = BankAccount("198777", 1000);
BAObject2 = BankAccount("156669", 1000);
BAObject2.print_account_info()
BAObject1.print_account_info()
BAObject1.deposit(500)
BAObject1.print_account_info()   
BAObject1.withdraw(200)
BAObject1.print_account_info()
print(f"Current Balance: {BAObject1.check_balance()}")
```

```python
class dad:
    def dadProperty(this):
        print("This is a property of the dad class")
    def commonProperty(this):
        print("This is a common property of the dad class")
class mom:
    def momProperty(this):
        print("This is a property of the mom class")
    def commonProperty(this):
        print("This is a common property of the mom class")

class child(mom, dad):
    def commonProperty(this):
        print("This is a common property of the child class")
childObj = child()
childObj.dadProperty()
childObj.momProperty()
childObj.commonProperty()
print(child.mro())  # This will show the method resolution order, which is child -> mom -> dad
```

```python
class grandparent:
    def grandparentProperty(this):
        print("This is a property of the grandparent class")
    def commonProperty(this):
        print("This is a common property of the grandparent class")
class parent(grandparent):
    def parentProperty(this):
        print("This is a property of the parent class")
    def commonProperty(this):
        print("This is a common property of the parent class")
class child(parent):
    def childProperty(this):
        print("This is a property of the child class")
    def commonProperty(this):
        print("This is a common property of the child class")
childObj = child()
childObj.grandparentProperty()
childObj.parentProperty()
childObj.childProperty()
childObj.commonProperty()
print(child.mro())
```

```python
# method overloading
import multipledispatch
class Calculator:
    @multipledispatch.dispatch(int, int)
    def add(a, b):
        return a + b
    @multipledispatch.dispatch(int, int, int)  
    def add( a, b, c):
        return a + b + c
calc = Calculator()
print(calc.add(2, 3))  # This will raise a TypeError since the second add method overwrites the first one
print(calc.add(2, 3, 4))  # This will work and return
```

```python
class Calculator1:
    
    def add(self,*args):
        if args :
          sum = type(args[0])()
          for num in args:
            sum += num
          return sum
        
calc1 = Calculator1()
print(calc1.add(2, 3))  # This will raise a TypeError since the second add method overwrites the first one
print(calc1.add(2, 3, 4))  # This will work and return
```

```python
class Parent:
    publicData = 10
    _protectedData = 20
    def publicMethod(this):
        print("This is a public method of the parent class " + str(Parent.publicData) )
class Child(Parent):
    def childMethod(this):
        print("This is a method of the child class")

parentObj = Parent()
print(parentObj.publicData)  # Accessing public data directly
parentObj.publicMethod()  # Accessing public method directly
print('========================================')
childObj = Child()
print(childObj.publicData)  # Accessing inherited public data directly
childObj.publicMethod()  # Accessing inherited public method directly
```

```python
class Car:
    def __init__(this, make, model):
        this.make = make
        this.__model = model
    def display_info(this):
        print(f"Car Make: {this.make}, Model: {this.__model}")
car1 = Car("Toyota", "Camry")
car2 = Car("Honda", "Civic")
car1.display_info()
car2.display_info()
class SubCar(Car):

    def __init__(this, make, model):
        super().__init__(make, model)

subcar1 = SubCar("Ford", "Mustang")
subcar1.display_info()
```
