```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [100, 120, 150, 130, 170, 200, 220, 210, 190, 160, 140, 110]
size = [100, 120, 150, 130, 170, 200, 220, 210, 190, 160, 140, 110] 
color = ['red', 'orange', 'yellow', 'green', 'blue', 'indigo', 'violet', 'pink', 'cyan', 'magenta', 'brown', 'gray']
plt.scatter(months, sales, s= np.array(size) , c=color, cmap='viridis', alpha=0.2, edgecolors='brown', linewidth=3)
#plt.colorbar()
plt.colorbar().set_label("Sales Volume")
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('scatter_plot.png')
```
