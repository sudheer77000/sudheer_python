```python
# Bar chart using Matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

overs =  np.arange(1, 21)
print(overs)
runs = np.random.randint(0, 37, 20)
print(runs)
wickets = np.random.randint(0, 2, 20)
print(wickets)

width = 0.25

plt.xticks(np.arange(1, 21, 1))
plt.yticks(np.arange(0, 36, 5))
plt.bar(overs + width, runs)
plt.bar(overs - width, wickets)
plt.legend(['Runs', 'Wickets'])



```

```python

plt.subplot(1,2,1)
plt.bar(overs + width, runs)
plt.subplot(1,2,2)
plt.bar(overs - width, wickets)
```
