```python
class parent:
   publicData = 10
   _protectedData = 20
   __privateData = 30
   def publicMethod(self):
       print('============= Public Method Calling Start =============')
       print('publicMethod with public data : ' + str(self.publicData))
       print('publicMethod with protected data : ' + str(self._protectedData))
       print('publicMethod with private data : ' + str(self.__privateData))
       print('============= Public Method Calling End =============')
   def _protectedMethod(self):
       print('============= Protected Method Calling Start =============')
       print('protectedMethod with public data : ' + str(self.publicData))
       print('protectedMethod with protected data : ' + str(self._protectedData))
       print('protectedMethod with private data : ' + str(self.__privateData))
       print('============= Protected Method Calling End =============')
   def __privateMethod(self):
       print('============= Private Method Calling Start =============')
       print('privateMethod with public data : ' + str(self.publicData))
       print('privateMethod with protected data : ' + str(self._protectedData))
       print('privateMethod with private data : ' + str(self.__privateData))
       print('============= Private Method Calling End =============')

class child(parent):
   def childMethod(self):
       print('childMethod with public data : ' + str(self.publicData))
       print('childMethod with protected data : ' + str(self._protectedData))
       #print('childMethod with private data : ' + str(self.__privateData))

print('############Parent Object Calling############')
p = parent()

print("Parent Public Data Variable: " + str(p.publicData))
p.publicMethod()

print("Parent Protected Data Variable: " + str(p._protectedData))
p._protectedMethod()

print("Parent Private Data Variable: " + str(p._parent__privateData))
p._parent__privateMethod()


print('############Child Object Calling############')

c = child()

print("Child Public Data Variable: " + str(c.publicData))
c.publicMethod()

print("Child Protected Data Variable: " + str(c._protectedData))
c._protectedMethod()

print("Child Private Data Variable: " + str(c._parent__privateData))
c._parent__privateMethod()

print(dir(p))
```
