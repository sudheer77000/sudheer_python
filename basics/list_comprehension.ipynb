```python
def lst_square(lst):
    for i in lst:
        lst.append(i**i)

my_list = [1, 2, 3]
lst_square(my_list)
```

```python
my_list = [1, 2, 3]
[i**i for i in my_list]

# for even numbers only
my_list = [1, 2, 3]
[i**i for i in my_list if i % 2 == 0]

# for odd numbers only
my_list = [1, 2, 3]
[i**i for i in my_list if i % 2 != 0]
```
