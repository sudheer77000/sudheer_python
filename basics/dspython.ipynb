```python
#Boolean  Operators
myString = "Sudheer_123"
print(myString)
print(myString.isalnum())
print(myString.isalpha())
print(myString.isdigit())
print(myString.islower())
print(myString.isupper())
print(myString.count("e"))
print(myString.startswith("s"))
print(myString.istitle())
```

```python
#Logical Operators with Variables
F1 = "Sudheer"
F2 = "Reddy"
print(F1.isalnum() and F2.isalnum())
```

```python
#List
myList = [1, 2, 3, 4, 5]
print(myList)
print(type(myList))
print(len(myList))
```

```python
#List Advanced
empList = ["Sudheer", "Reddy", 30, "Engineer", 75000,True,11.369]
print(empList)
empList.append("Python")
print(empList) 
empList.insert(-2,"Kumar")
print(empList)
empList.extend(["Python","Hari"])
print(empList)
print(empList[5])
print(empList[-1])
```

```python
#Set 
mySet = {5,4,5,7,1,4}
stingSet = {"Sudheer", "Reddy", "Sudheer"}
print(stingSet)
print(mySet)
mySet.add('r')
print(mySet)
```

```python
# Set Difference
setA = {1, 2, 3, 4, 5}
setB = {4, 5, 6, 7, 8}
difference = setB.difference(setA)
print(difference)
setA.difference_update(setB)
print(setA)
```

```python
# Dictionary
myDict = {"Name": "Sudheer", "Age": 30, "City": "Hyderabad"}
print(myDict)
print(myDict["Name"])
myDict["Age"] = 31
print(myDict)
print(myDict.keys())
print("=====================")
for key in myDict:
    print(key)
print("=====================")
for key, value in myDict.items():
    print(f"{key}: {value}")
    print(key)
    print(value)
```

```python
# Nested Dictionary
sudheeerDict = {"Name": "Sudheer", "Age": 30}
ReddyDict = {"Name": "Reddy", "Age": 25}
nestedDict = {
    "Person1": sudheeerDict,
    "Person2": ReddyDict
}
print(nestedDict)
```
