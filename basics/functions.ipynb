```python
# Even Or Odd
num = 498882
if num % 2 == 0:
    print(f"{num} is an even number.")
else:
    print(f"{num} is an odd number.")
```

```python
def is_even_or_odd(num):
    if num % 2 == 0:
        return f"{num} is an even number."
    else:
        return f"{num} is an odd number."
```

```python
is_even_or_odd(989)
```

```python
# print vs return
def hello_message(name):
    print(f"Hello, {name}!")
    return f"Hello, {name}!!"  # This will return the string instead of printing it.

message = hello_message("Sudheer")
print(message)  # This will print None because the function does not return anything.
```

```python
# add 2 numbers
def add_numbers(a, b):
    return a + b
result = add_numbers(5698, 805)
print(result)  # Output: 8
```

```python
# positional vs keyword arguments
def greet(name, greeting):
    return f"Hello, {name}! {greeting}"
print(greet("Sudheer", "Welcome to the world of Python!"))  # Positional arguments
print(greet(greeting="Welcome to the world of Python!", name="Sudheer"))  # Keyword arguments
```

```python
# Default arguments
def greet(name, greeting="Welcome"):
    return f"Hello, {name}! {greeting}"
print(greet("Sudheer","Welcome to India"))  # Uses default greeting
```

```python
# arbitory arguments
def greet(*names):
    return f"Hello, {', '.join(names)}!"
print(greet("Sudheer", "Alice", "Bob","ArithmeticError","Charlie"))  # Output: Hello, Sudheer, Alice, Bob!
def sum_numbers(*numbers):
    return sum(numbers) # This will sum all the numbers passed as arguments.
print(sum_numbers(1, 2, 3, 4, 5))  # Output: 15
```

```python
# keyword arbitory arguments
def greet(**kwargs):
    return f"Hello, {', '.join(kwargs.keys())}!"
print(greet(name1="Sudheer", name2="Alice", name3="Bob"))  # Output: Hello, Sudheer, Alice, Bob!
```

```python
#arbitrary arguments and keyword arguments together
def greet(*names, **kwargs):
   print(f"Keyword arguments: {kwargs}")
   print(f"Positional arguments: {names}") 
greet("Sudheer", "Alice", "Bob","Anu", greeting="Welcome",test="suma")  # Output: Welcome, Sudheer, Alice, Bob!
```

```python
# lambda functions
add = lambda x, y: x + y
print(add(5, 3))  # Output: 8
```

```python
# even odd using lambda
is_even = lambda x: f"{x} is an even number." if x % 2 == 0 else f"{x} is an odd number."
print(is_even(498882))  # Output: 498882 is an even number.
```

```python
# lambda passing 3 arguments
add_three = lambda x, y, z: x + y + z
print(add_three(1, 2, 3))  # Output: 6
```

```python
# map function even odd
numbers = [1, 2, 3, 4, 5,10,12,13,14,15,19,20]
list(map(is_even_or_odd, numbers))
```
