```python
from io import StringIO, BytesIO
import numpy as np
import pandas as pd
```

```python
data = ('col1,col2,col3\n1,2,3\n4,5,6\n7,8,9')
type(data)
print(data)
```

```python
pd.read_csv(StringIO(data))
```

```python
# Get Specific Column
pd.read_csv(StringIO(data), usecols=['col2', 'col3'])
```

```python
# Get specific rows
pd.read_csv(StringIO(data), skiprows=1, nrows=2)
```

```python
data = 'Name,Age,City\nAlice,30,New York\nBob,25,Los Angeles\nCharlie,35,Chicago'
print(data)
```

```python
df = pd.read_csv(StringIO(data),dtype='object')
#print(df)
print(df['Age'])
```

```python
df = pd.read_csv(StringIO(data),dtype={'Age': float})
print(df['Age'])
print(df.dtypes)
```
