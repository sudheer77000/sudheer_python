```python
from abc import ABC, abstractmethod
class BaseModel(ABC):
    @abstractmethod
    def predict(self, input_data):
        pass
    def print_model_info(self):
        print("This is a base model.")
    @abstractmethod
    def train(self, training_data):
        pass

#base_model = BaseModel()  # This will raise an error because BaseModel is abstract

class LinearModel(BaseModel):
    def predict(self, input_data):
        return [x * 2 for x in input_data]  # Dummy implementation
    def print_model_info(self):
        print("This is a linear model.")
    def train(self, training_data):
        print("Training linear model with data:", training_data)

linear_model = LinearModel()
linear_model.print_model_info()  # Output: This is a linear model.
linear_model.predict([1, 2, 3])  # Output: [2, 4, 6]
```
