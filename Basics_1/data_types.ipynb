```python
# Dictionary
d = {'name': "Sudheer", "age": 30, "city": "Hyderabad"}
print(d)
print(type(d))
```

```python
# Dictionary Duplicate Keys
d = {"name": "Sudheer", "age": 30, "city": "Hyderabad", "name": "John"}
print(d)
```

```python
#Dictionary Keys Methods
d = {"name": "Sudheer", "age": 30, "city": "Hyderabad"}
print(d.keys())
print(type(d.keys()))
print(d.values())
print(type(d.values()))
print(d.items())
print(type(d.items()))
for i,j in d.items():
    print("Key:", i, ",Value:", j)
print("================")
d.update({"name": "John"})
d["continent"] = "Asia"
print(d)
d.update({"city": "Dubai", "grade": "5"})
print(d)
```

```python
# List
l = []
print(l)
print(type(l))
l = [1, 2, 3,"Sudheer", 4,"Gundra", 5, True, None,15.5, 2 + 3j]
print(l)
# Append
l.append("Python")
print(l)
l.extend([6, 7, 8,"Honey", 1, 4])
print(l)
# Remove
l.remove("Sudheer")
print(l)
l.remove(1)
print(l)
#pop
l.pop()
print(l)
l.pop(0)
print(l)
print("Count of List:", len(l))
print("Count of 4 in List:", l.count(4))
#insert
l.insert(5, "Sudheer")
print(l)
# delete
del l[5]
print(l)
```
