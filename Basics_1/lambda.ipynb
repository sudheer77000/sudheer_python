```python
def add(x,y):        return x+y
def subtract(x,y):    return x-y
def multiply(x,y):    return x*y
def divide(x,y):    return x/y
print(add(10,5))
print(subtract(10,5))
print(multiply(10,5))
print(divide(10,5))
# Lambda functions
add1 = lambda x,y: x+y
print(add1(10,5))
# Lambda function Square
square = lambda x: x*x
print(square(10))
# Declare and Call Lambda function
print((lambda x: x*x)(6))
# Lambda function with if-else
(lambda x,y: "X is greater" if x>y else "Y is greater")(10,5)
(lambda name : name + " is longer" if len(name)>5 else name + " is shorter")("Arun")
(lambda x: "A" if x >= 90 else "B" if x >= 75 else "C" if x >= 60 else "F")(60) #lambda name : name + " is longer" if len(name)>5 else name + " is shorter"
```
