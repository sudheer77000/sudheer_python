```python
# Class Methods
class MyClass:
    class_variable = "I am a class variable"
    
    # Instance method
    def instance_method(self):
        print('This is a Instance Method')
        print(self.class_variable)

    @classmethod
    def class_method(self):
        print('This is a Class Method')
        print(self.class_variable)

    @staticmethod
    def static_method():
        print('This is a Static Method')
        print(MyClass.class_variable)

    @staticmethod
    def updateClassVar():
        MyClass.class_variable = 'I am overrided'
        #del MyClass.class_variable

myclassObj = MyClass()
myclassObj.instance_method()
myclassObj.class_method()
myclassObj.static_method()
myclassObj.updateClassVar()
print(myclassObj.class_variable)

myclassObj2 = MyClass()
print(myclassObj2.class_variable)
```

```python
class Employee:
    company = "ABC Ltd"

    def __init__(self, name):
        self.name = name

    # Instance Method
    def show_name(self):
        print(self.name)

    # Class Method
    @classmethod
    def show_company(cls):
        print(cls.company)

    # Static Method
    @staticmethod
    def greet():
        print("Welcome")
```

```python
emp = Employee("Sudheer")
emp.show_name()  
#Employee.show_name()       
#Employee.show_company()  
#Employee.greet()     
```
