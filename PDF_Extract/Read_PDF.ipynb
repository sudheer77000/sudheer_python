```python
import fitz
pdf_path = "3_Page_Doc.pdf"
doc = fitz.open(pdf_path)
```

```python
print("Number of pages in the PDF:", doc.page_count)
print("Metadata of the PDF:", doc.metadata)
print("Table of contents of the PDF:", doc.get_toc())
print("Page labels of the PDF:", doc.get_page_labels())
print("Page layout of the PDF:", doc.chapter_count)
text = ""
for page in doc:
    text += page.get_text()

print("Text extracted from the PDF:", text)
doc.close()
```
