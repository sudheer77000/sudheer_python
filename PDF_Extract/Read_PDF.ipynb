```python
import fitz
pdf_path = "3_Page_Doc.pdf"
doc = fitz.open(pdf_path)
print("Number of pages in the PDF:", doc.page_count)
print("Metadata of the PDF:", doc.metadata)
print("Table of contents of the PDF:", doc.get_toc())
print("Page labels of the PDF:", doc.get_page_labels())
print("Page layout of the PDF:", doc.chapter_count)
text = ""
print("Number of pages:", len(doc))
page = doc.load_page(1)
print("=========================")
print(page.get_text())
print("=========================")
#for page in doc:
#    text += page.get_text()
#
#print("Text extracted from the PDF:", text)
#doc.close()
```

```python
all_text = ""

for page in doc:
    all_text += page.get_text()

print(all_text)
```

```python
text = page.get_text("dict")
print(text)
```

```python
blocks = page.get_text("blocks")
print(blocks)
```

```python
rects = page.search_for("Data")

for rect in rects:
    print(rect)
```

```python
images = page.get_images(full=True)

print(images)
for img_index, img in enumerate(images):

    xref = img[0]   # Image reference number

    image_data = doc.extract_image(xref)

    image_bytes = image_data["image"]
    image_ext = image_data["ext"]

    filename = f"image_{img_index}.{image_ext}"

    with open(filename, "wb") as f:
        f.write(image_bytes)

    print("Saved:", filename)
```
