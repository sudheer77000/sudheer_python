```python

import langchain
print("LangChain installed successfully")
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from langchain_openai import OpenAIEmbeddings
api_key = os.getenv("OPENAI_API_KEY")
OpenAIEmbeddingModel = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=api_key
)
my_text = "Sudheer Gundra"
my_text_embedding = OpenAIEmbeddingModel.embed_query(my_text)
print(my_text_embedding)
print(len(my_text_embedding))
```
