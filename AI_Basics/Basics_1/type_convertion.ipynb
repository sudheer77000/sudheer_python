```python
# Type Convertion
a = 1
b = 2.9
c = "3"
d = "Sudheer"
e = True
f = [1,2,3,4,5]
g = a + b
print(g) # Automatically converts a to float and adds it to b
print(type(g))
# Common functions int(), float(), str(), bool(), list(), tuple(), set(), dict()
h = str(a) # Converts a to string
print(h)
print(type(h))
print(int(b)) # Converts b to integer (truncates the decimal part)
print(float(c)) # Converts c to float
print(bool(e)) # Converts e to boolean (True)
print(set(f)) # Converts f to a set (removes duplicates)
```
