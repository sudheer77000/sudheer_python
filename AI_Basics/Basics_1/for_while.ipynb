```python
# For
import time
set_1 = {1, 2, 3, 4, 5}
for i in set_1:
    time.sleep(1)
    print(i)
print("Done")
for i in range(1, 11):
    time.sleep(1)
    print(i)
print("Done")
b = "Sudheer Gundra"
for k in b:
    time.sleep(1)
    print(k)
```
