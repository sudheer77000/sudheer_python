```python
from sentence_transformers import SentenceTransformer

# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Input sentence
sentence = "Sudheer is working in Dubai"

# Generate embedding
embedding = model.encode(sentence)

# Output
print("\nSentence:", sentence)
print("\nEmbedding size:", len(embedding))

print("\nEmbedding values (first 20):")
print(embedding[:20])
```

```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("sentence-transformers/all-MiniLM-L6-v2")
model = AutoModel.from_pretrained("sentence-transformers/all-MiniLM-L6-v2")

text = "My name is John. I live in New York."

inputs = tokenizer(text, return_tensors="pt")
tokens = tokenizer.tokenize(text)
print(tokens)

with torch.no_grad():
    outputs = model(**inputs)

print(outputs.last_hidden_state.shape)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "sentence-transformers/all-MiniLM-L6-v2"
)

text = "Sudheer working as Software Engineer. Sudheer loves watching football."

# Encode text
encoded = tokenizer(text)

# Convert IDs back to tokens
tokens = tokenizer.convert_ids_to_tokens(encoded["input_ids"])

print("Total Tokens:", len(tokens))
print("\nTokens:")
for i, token in enumerate(tokens):
    print(f"{i:2d}: {token}")
```
