```python
# Normal List
list1 = [1, 2, 3, 4, 5]
print("Normal List:", list1)
print(list1*5)
list2 = []
for i in list1:
    list2.append(i*5)
print("List Comprehention:", list2)

# List Comprehension
list3 = [i*5 for i in list1]
print("List Comprehention:", list3)

list4 = list(range(1, 6))
list5 = [i for i in list4 if i % 2 == 0]
print("Even Numbers:", list5)
```
