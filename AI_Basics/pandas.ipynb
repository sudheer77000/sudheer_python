```python
import pandas as pd
import numpy as np

print(pd.__version__)
print(np.__version__)
```

```python
# Major Data Structure
# 1. Series - One Dimentional Array
# 2. DataFrame - Two Dimentional Array
# 3. Panel - Three Dimentional Array

a = pd.Series([1,2,3,4,5])
print(a)
b = pd.Series([1,2,3,4,5], index=['e','f','g','h','i'])
print(b)

```

```python
# Describe

print(a.describe())
print(a.min())
```

```python
# Data Manuplation
doubled = a.map(lambda x: x*2)
print(doubled)
b = pd.Series([5,2,3,8,4,2,7,9,8,9])
a = b.sort_values()
print(a) 
```

```python
# inplace & Dropped
s1 = pd.array([1,2,3,4,5])
print(s1)
dropped_series = s1.delete(0);
print(dropped_series)
```

```python
# Handling Missing Data
s4 = pd.Series([1, 2, 3, 4, 5, 6, np.nan])

print(s4)
print(s4.isnull())
print(s4.notnull())  # This will work perfectly now!
print(s4.fillna(0))
print(s4.dropna())

```

```python
# Indexing and slicing
s6 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
print(s6)
print(s6.iloc[4])
print(s6.iloc[1:3])
print(s6.loc['e'])
print(s6.loc['b':'d'])
```

```python
# Filter And Aggirigation
s7 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
print(s7[s7 > 2])
print(s7.sum())
print(s7.cumsum())
print(s7.aggregate(['sum','mean','std'])) # One Or More Operation
```

```python
# Grouping
genre_groups = df.groupby('genre')
genre_groups
```

```python
for genre, group_data in genre_groups:
    print(f"Genre: {genre}")
    print(group_data)
    print()
```

```python
# Average Duration
avg_duration_by_genre = df.groupby('genre')['duration'].mean()
avg_duration_by_genre
```

```python
# Group By multiple columns
stats_by_genre = df.groupby('genre')['avg_vote'].agg([np.mean,np.std,np.min,np.max])
stats_by_genre
```

```python
# Complex Aggeregation
complex_aggregation = df.groupby('genre').agg({
    'duration' : np.mean,
    'avg_vote' : [np.min,np.max],
    'public_vote' : 'sum'
})
complex_aggregation
```
