```python
from sentence_transformers import SentenceTransformer
```

```python
sentence_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentence_embeddings = sentence_model.encode(["Hello, my dog is cute", "What is the capital of France?"])
print(sentence_embeddings)
sentence_embeddings.shape
print(len(sentence_embeddings))
```

```python
my_paragraph = """The cat is on the table.
The dog is in the garden.
The bird is flying in the sky.
The fish is swimming in the water.
The horse is running in the field."""
paragraph_embeddings = sentence_model.encode(my_paragraph)
print(paragraph_embeddings)
paragraph_embeddings.shape
```
