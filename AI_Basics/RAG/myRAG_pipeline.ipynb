```python
from pathlib import Path
BASE_DIR = Path.cwd()
PDF_File_COMPLEX = BASE_DIR / "fab_credit_card_statement_june_2026.pdf"
print(PDF_File_COMPLEX)
file_path= PDF_File_COMPLEX
```

```python
pages=PyPDFLoader(file_path).load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
chunk=RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=400)
```

```python
chunk_data = chunk.split_documents(pages)
```

```python
len(chunk_data)
```

```python
import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
```

```python
index = faiss.IndexFlatL2(1536)
```

```python
vector_store=FAISS(embedding_function=embedding,index=index,docstore=InMemoryDocstore(), index_to_docstore_id={} )
```

```python
vector_store.add_documents(chunk_data)
```

```python
retriever=vector_store.as_retriever(search_kwargs={"k":10})
```

```python
retriever.invoke("what is name of credit card holder")
```

```python
retriever.invoke("what is name of credit card holder")[0].page_content
```

```python
template= """ 

Answer  the Question based on the following  {context}

Question:{question}
"""
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
promte=PromptTemplate.from_template(template)
```

```python
promte
```

```python
def format_doc(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.runnables import  RunnablePassthrough
```

```python
rag_chain= {"context": retriever|format_doc, "question": RunnablePassthrough() }|promte| myllm | StrOutputParser()
```

```python
rag_chain.invoke("what is name of credit card holder?")
```

```python
rag_chain.invoke("Highest amount spent in june 2026?")
```
