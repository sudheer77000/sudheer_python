```python
import os
from dotenv import load_dotenv
from llama_cloud_services import LlamaParse

load_dotenv()

api_key = os.getenv("LLAMA_PARSE_KEY")

parser = LlamaParse(
api_key=api_key,
result_type="markdown"
)

documents = parser.load_data("doc.pdf")

for document in documents:
    print(document.text)
```
