```python
# Create Numpy Array
# From list
Arr1 = np.array([1,2,3,4,5])
print(Arr1)
print(type(Arr1))
#Default Functions
zero = np.zeros((3,3))
print(zero)
one = np.ones((4,2))
print(one)
ar = np.arange(10)
print(ar)
ar_even = np.arange(0,10,2)
print(ar_even)
ar_odd = np.arange(1,10,2)
print(ar_odd)
lp = np.linspace(0,2,5)
print(lp)
iden = np.eye(4)
print(iden)

```

```python
# Randon
randf = np.random.rand(2,3)
print(randf)

randin = np.random.randint(1,100,size=(3,3,3))
print(randin)
```

```python
print(np.arange(2,11,2).astype(int))
print(np.ones((3,3)).astype(int))
print(np.arange(2,11,2).astype(str))
```

```python
# Attributes .shape,.size,ndim
arr = np.array([[[10,20,30],
                [40,50,60],
                [70,80,90]]])
print(arr)
print(arr.shape)
print(arr.size)
print(arr.ndim)
```

```python
# Indexing and Slicing
print(arr[0,2,0])
print(arr[0:,0:3,0:3])
```

```python
# Mathamatical operations
a = np.array([1,2,3])
b = np.array([1,2,4])
print(a + b)
print(a * b)
print(a -b)
print(a/b)
print(np.sqrt(a))
```

```python
# Array Methods
# min, max , mean, median, std, axis operation
a = np.arange(0,100,6)
print(a)
print(np.min(a))
print(np.max(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
b = np.array([[1,2,3],
              [5,7,8]])
print(np.sum(b))
print(np.sum(b,axis=0))
print(np.sum(b,axis=1))
```

```python
# Broadcasting and reshaping
arr1 = np.array([[1,2,3],
                [4,5,6]])
print(arr1)
print(arr1 + 10)
```

```python
# Reshape
arr2 = np.arange(0,12)
print(arr2)
print(arr2.reshape(2,6,1))
print(arr2.reshape(2,6,1,1).ndim)
```

```python
a = np.array([1,2,3])
b = np.array([1,2,4])
print(np.concatenate((a,b)))
print(np.vstack((a,b)))
print(np.hstack((a,b)))
c = np.arange(1,100)
print(c)
print(np.sort(b))
print(np.sort(b)[::-1])
```

```python
# Advance Tricks
d = np.arange(1,10)
print(d[d > 5])
e = [1,2,3,12,4,2,9]
print(np.unique(e))
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = np.matmul(A, B)
D = np.matmul(B, A)
# or C = A @ B

print(C)
print(D)
```
