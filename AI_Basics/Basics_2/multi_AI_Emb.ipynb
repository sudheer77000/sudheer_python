```python
import os
from dotenv import load_dotenv
load_dotenv()
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
#print(GEMINI_API_KEY)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
myEmbedding = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")
len(myEmbedding.embed_query("Total Number of building in UAE"))
```

```python
from langchain_openai import OpenAIEmbeddings
api_key = os.getenv("OPENAI_API_KEY")
OpenAIEmbeddingModel = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=api_key
)
my_text = "Sudheer, hello"
my_text_embedding = OpenAIEmbeddingModel.embed_query(my_text)
print(my_text_embedding)
print(len(my_text_embedding))
```
