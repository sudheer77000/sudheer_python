```python
import time

def task(name):
    print(f"Starting {name}")
    time.sleep(2)  # Simulate work
    print(f"Finished {name}")

task("Task 1")
task("Task 2")
task("Task 3")
task("Task 4")
task("Task 5")
```

```python
import asyncio

async def task(name):
    print(f"Starting {name}")
    await asyncio.sleep(2)  # Non-blocking wait
    print(f"Finished {name}")

async def main():
    await asyncio.gather(
        task("Task 1"),
        task("Task 2"),
        task("Task 3"),
        task("Task 4"),
        task("Task 5"),
    )
    
await main()
```
