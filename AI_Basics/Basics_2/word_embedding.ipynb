```python
from gensim.models import Word2Vec, KeyedVectors
import gensim.downloader as api
# skip-gram model trained on Google News dataset
model = api.load("word2vec-google-news-300")
```

```python
vector = model["maha"]
print(vector)
print(len(vector))
```

```python
print(model.similar_by_word("king", topn=10))
```

```python
print(model.most_similar("cricket", topn=10))
```

```python
print(model.most_similar("apple", topn=10))
```

```python
similarity = model.similarity("car", "cars")
print(similarity)
```

```python
similarity = model.similarity('democracy', 'banana')
print(similarity)
```

```python
similarity = model.similarity('car', 'audi')
print(similarity)
```

```python
result = model.most_similar(
    positive=['apple', 'orange','technology'],
    negative=['fruits'],
    topn=5
)

print(result)
```

```python
word = "sudheer"

if word in model:
    print("Word exists")
else:
    print("Word not found")
```

```python
words = ["apple", "banana", "orange", "grapes"]

odd = model.doesnt_match(words)
print(odd)
```
