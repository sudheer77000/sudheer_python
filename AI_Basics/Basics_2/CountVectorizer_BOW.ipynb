```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Sample documents
documents = [
    "movie is excellent.Hero action is excellent",
    "movie is very bad",
    "excellent acting",
    "bad story"
]

# Create CountVectorizer object
cv = CountVectorizer()
X = cv.fit_transform(documents)
df = pd.DataFrame(
    X.toarray(),
    columns=cv.get_feature_names_out()
)

print(df)
```
