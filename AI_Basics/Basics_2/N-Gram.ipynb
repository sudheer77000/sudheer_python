```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love learning Python",
    "Python is easy to learn"
]

cv = CountVectorizer(ngram_range=(4, 4))

X = cv.fit_transform(documents)

print("Features:")
print(cv.get_feature_names_out())

print("\nMatrix:")
print(X.toarray())
```
