```python
from sentence_transformers import SentenceTransformer, util

# 1. Load a lightweight, highly efficient semantic transformer model
# 'all-MiniLM-L6-v2' is a fast, standard model for semantic search
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Define our source sentence (the query)
query = "What time does the Champions League final kick off tonight?"

# 3. Define candidate sentences to compare against
candidates = [
    "Show me the updated point table for the Indian Premier League.", # Semantically highly similar
    "Who won the men's singles finals at Wimbledon yesterday?",                # Completely unrelated
    "Can you check the current live score for the basketball game?",    # Somewhat related (AI context)
]

# 4. Encode the text into dense vector embeddings (semantic representations)
query_embedding = model.encode(query, convert_to_tensor=True)
candidate_embeddings = model.encode(candidates, convert_to_tensor=True)

# 5. Compute cosine similarity between the query and all candidates
cosine_scores = util.cos_sim(query_embedding, candidate_embeddings)[0]

# 6. Display the results
print(f"Query: '{query}'\n")
print("Semantic Similarity Scores:")
print("-" * 30)

for i, score in enumerate(cosine_scores):
    print(f"Score: {score.item():.4f} | Match: '{candidates[i]}'")
```
