```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(0, 100, 20).reshape(5, 4),index=['R1', 'R2', 'R3', 'R4' , 'R5'], columns=['C1', 'C2', 'C3', 'C4'])
print(df)
df.to_csv('data.csv', index=True, header=True)
print("=================================")
print(df.loc['R1'])
print("=================================")
print(df['C1'])
print("=================================")
print(df.loc['R4':'R5', 'C3':'C4'])
```

```python
df = pd.DataFrame(np.random.randint(0, 100, 20).reshape(5, 4),index=['R1', 'R2', 'R3', 'R4' , 'R5'], columns=['C1', 'C2', 'C3', 'C4'])
print(df)
print("=================================")
print(df.iloc[1:2,0:2])
print("=================================")
print(df.loc['R2':'R3', 'C1':'C2'])
print("=================================")
print(df.iloc[0:2,1:4])
print("=================================")
print(df.loc['R1':'R3', 'C1':'C4'])
```
