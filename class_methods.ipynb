```python
class owner:
    Name = "Deccan Charges"
    Location = "Hyderabad"
    def printDetails(self):
        print(f"Owner Name: {self.Name}")
        print(f"Owner Location: {self.Location}")
    @classmethod
    def changeOwnerName(self, new_name):
        self.Name = new_name    
owner1 = owner()
owner1.printDetails()
owner1.changeOwnerName("Sun Risers Hyderabad")
owner1.printDetails()

owner2 = owner()
owner2.printDetails()

owner.printDetails(owner2)
```
